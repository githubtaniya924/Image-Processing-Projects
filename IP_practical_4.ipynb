```python
from PIL import Image, ImageFilter
from skimage import data
import numpy as np
import matplotlib.pyplot as plt

#load built-in satellite image from scikit-image
satellite_img = data.hubble_deep_field()

#convert numpy array to PIL image
img = Image.fromarray(satellite_img)

median_img = img.filter(ImageFilter.MedianFilter(size=3))

#apply smoothing (blur)
smooth_img=img.filter(ImageFilter.BLUR)

#appy sharpening
sharp_img = img.filter(ImageFilter.SHARPEN)

#save the outout
median_img.save('satellite_median_filtered.jpg')
smooth_img.save('satellite_smoothed.jpg')
sharp_img.save('satellite_sharpened.jpg')

#Display all images
plt.figure(figsize=(12,8))

plt.subplot(2,2,1)
plt.title('Original Satellite Image')
plt.imshow(img)
plt.axis("off")

plt.subplot(2,2,2)
plt.title('median Satellite Image')
plt.imshow(median_img)
plt.axis("off")

plt.subplot(2,2,3)
plt.title('smoothed Satellite Image')
plt.imshow(smooth_img)
plt.axis("off")

plt.subplot(2,2,4)
plt.title('sharpened Satellite Image')
plt.imshow(median_img)
plt.axis("off")

```

```python
from PIL import Image, ImageFilter
from skimage import data
import numpy as np
import matplotlib.pyplot as plt

import imageio.v2 as imageio
satellite_img = imageio.imread("img3.jpg")

#load built-in satellite image from scikit-image
#satellite_img = data.hubble_deep_field()

#convert numpy array to PIL image
img = Image.fromarray(satellite_img)

median_img = img.filter(ImageFilter.MedianFilter(size=3))

#apply smoothing (blur)
smooth_img=img.filter(ImageFilter.BLUR)

#appy sharpening
sharp_img = img.filter(ImageFilter.SHARPEN)

#save the outout
median_img.save('satellite_median_filtered.jpg')
smooth_img.save('satellite_smoothed.jpg')
sharp_img.save('satellite_sharpened.jpg')

#Display all images
plt.figure(figsize=(12,8))

plt.subplot(2,2,1)
plt.title('Original Satellite Image')
plt.imshow(img)
plt.axis("off")

plt.subplot(2,2,2)
plt.title('median Satellite Image')
plt.imshow(median_img)
plt.axis("off")

plt.subplot(2,2,3)
plt.title('smoothed Satellite Image')
plt.imshow(smooth_img)
plt.axis("off")

plt.subplot(2,2,4)
plt.title('sharpened Satellite Image')
plt.imshow(median_img)
plt.axis("off")

```

```python
from skimage.metrics import structural_similarity as ssim
from skimage.metrics import peak_signal_noise_ratio as psnr
from skimage.metrics import mean_squared_error
import numpy as np

# Convert PIL images to numpy arrays
orig = np.array(img)
median3 = np.array(img.filter(ImageFilter.MedianFilter(size=3)))
median7 = np.array(img.filter(ImageFilter.MedianFilter(size=7)))

# --- Compute Metrics ---
mse3 = mean_squared_error(orig, median3)
mse7 = mean_squared_error(orig, median7)

psnr3 = psnr(orig, median3)
psnr7 = psnr(orig, median7)

ssim3 = ssim(orig, median3, channel_axis=2)
ssim7 = ssim(orig, median7, channel_axis=2)

print("\n----- Statistical Comparison -----")
print(f"MSE (Median 3): {mse3}")
print(f"MSE (Median 7): {mse7}\n")

print(f"PSNR (Median 3): {psnr3}")
print(f"PSNR (Median 7): {psnr7}\n")

print(f"SSIM (Median 3): {ssim3}")
print(f"SSIM (Median 7): {ssim7}")
```

```python
import matplotlib.pyplot as plt

plt.figure(figsize=(12, 4))
plt.hist(orig.ravel(), bins=256, alpha=0.5, label="Original")
plt.hist(median3.ravel(), bins=256, alpha=0.5, label="Median 3")
plt.hist(median7.ravel(), bins=256, alpha=0.5, label="Median 7")
plt.legend()
plt.title("Pixel Intensity Histogram Comparison")
plt.show()
```
